# Custom

```python
# 15 January 2026 ka Actual Data
actual_test_url = "https://data.elexon.co.uk/bmrs/api/v1/datasets/FUELHH/stream?publishDateTimeFrom=2026-01-15T00:00:00Z&publishDateTimeTo=2026-01-15T23:59:59Z"

df_actual = pd.read_json(actual_test_url)

# Sirf WIND wala filter karke check karo
df_wind_actual = df_actual[df_actual['fuelType'] == 'WIND']
print("Actual Wind Rows:", len(df_wind_actual))
df_wind_actual.head()
```

```python
# 15 January 2026 ka Forecast Data
forecast_test_url = "https://data.elexon.co.uk/bmrs/api/v1/datasets/WINDFOR/stream?publishDateTimeFrom=2026-01-14T00:00:00Z&publishDateTimeTo=2026-01-15T00:00:00Z"

df_forecast = pd.read_json(forecast_test_url)
print("Forecast Rows:", len(df_forecast))
df_forecast.head()
```

```python
test_date = "2026-10-27"
# 1. October 27 Actuals fetch karo
actual_url = f"https://data.elexon.co.uk/bmrs/api/v1/datasets/FUELHH/stream?publishDateTimeFrom={test_date}T00:00:00Z&publishDateTimeTo={test_date}T23:59:59Z"
df_act = pd.read_json(actual_url)
df_wind = df_act[df_act['fuelType'] == 'WIND']
```

```python
forecast_url = "https://data.elexon.co.uk/bmrs/api/v1/datasets/WINDFOR/stream?publishDateTimeFrom=2026-10-26T00:00:00Z&publishDateTimeTo=2026-10-27T00:00:00Z"
df_fc = pd.read_json(forecast_url)
df_target_fc = df_fc[df_fc['startTime'].str.startswith(test_date)]
print(f"[{test_date}] Actual Wind Rows:", len(df_wind))
print(f"[{test_date}] Day-Ahead Forecast Rows:", len(df_target_fc))
df_wind[['settlementPeriod', 'startTime', 'generation']].head()
```
